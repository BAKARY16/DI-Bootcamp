**Défi quotidien : Intégration de NumPy, Pandas et Matplotlib**

Vous êtes analyste de données et travaillez sur un ensemble de données météorologiques mondiales. Votre tâche consiste à analyser les tendances de température et à visualiser les résultats.

*1. Préparation des données*

```python
# Importer les bibliothèque néccessaire
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generer les données
data = np.random.uniform(0, 100, 1000)

# Création du DataFrame
df = pd.DataFrame(data, index=pd.date_range(start='2023-01-01', periods=1000, freq='D'), columns=['Value'])

# generer un jeu de donnée synthetique avec Numpy pour les température moyennes mensuelles
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperatures = np.random.uniform(-5, 35, 12)
temperatures = np.round(temperatures, 2)  

city_data = np.column_stack((months, temperatures))



# Convertir en DataFrame pandas
city_df = pd.DataFrame(city_data, columns=['Month', 'Temperature (°C)'])

city_df
```

*2. Analyse des données*

```python
# Calculez la température moyenne annuelle en utilisant DataFrame.mean(axis)
temp_moy_annelle = city_df['Temperature (°C)'].mean()
print(f"La temperature moyenne annuelle est : {temp_moy_annelle} °C")
```

```python
# Trouvez la ville avec la température moyenne la plus élevée et la plus basse en utilisant les méthodes idxmax()et idxmin()
city_level = city_df["Temperature (°C)"].idxmax()
print(f"La ville avec la température moyenne la plus élevée est : {city_level}")
```

```python
# Trouver la ville avec la température moyenne la plus basse
city_level_min = city_df["Temperature (°C)"].idxmin()
print(f"La ville avec la température moyenne la plus basse est : {city_level_min}")
```

```python
# Calculez la température moyenne annuelle pour chaque ville
temp_city = city_df.groupby('Month')['Temperature (°C)'].mean()
print(temp_city)
```

```python
# Identifiez la ville ayant la température moyenne la plus élevée et la plus basse pour l'année.
city_level_max = temp_city.idxmax()
city_level_min = temp_city.idxmin()

print(f"La ville avec la température moyenne la plus élevée pour l'année est la ville : {city_level_max}")
print(f"La ville avec la température moyenne la plus basse pour l'année est la ville : {city_level_min}")
```

*3. Visualisation des données*

```python
# Visualisation des données
plt.figure(figsize=(10, 6))
plt.plot(city_df['Month'], city_df['Temperature (°C)'], marker='o')
plt.title('Température Moyenne Mensuelle')
plt.xlabel('Mois')
plt.ylabel('Température (°C)')
plt.xticks(city_df['Month'])
plt.grid()
plt.show()
```

Un bref rapport dans le cahier résumant vos conclusions, y compris la ville avec les températures moyennes les plus élevées et les plus basses et toute tendance intéressante observée dans les données.

- Après les traitements effectués, nous pouvons dire que la ville avec la température moyenne la plus élevée pour l'année est la ville : City 7
- La ville avec la température moyenne la plus basse pour l'année est la ville : City 3
- En observant les données, nous pouvons également noter que les températures varient considérablement d'un mois à l'autre, avec des pics de chaleur en été et des températures plus fraîches en hiver, ce qui est typique pour de nombreuses régions du monde.
