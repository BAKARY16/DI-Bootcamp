```python
import pandas as pd
import numpy as np

# Simulating stock prices for a non-existent stock (DVLPRS)
np.random.seed(0)  # For reproducibility

# Generating 100 days of stock prices
dates = pd.date_range(start='2023-01-01', periods=100)
prices = np.random.normal(loc=100, scale=10, size=len(dates))  # Assume a mean price of 100 with a standard deviation of 10

# Creating a DataFrame
stock_data = pd.DataFrame({
    'Date': dates,
    'DVLPRS_Price': prices
})

# Displaying the first few rows of the simulated data
print(stock_data.head())
```

```python
# Calculez la moyenne, la médiane, la variance et l'écart type des cours boursiers
data_moy = stock_data['DVLPRS_Price'].mean()
data_median = stock_data['DVLPRS_Price'].median()
data_variance = stock_data['DVLPRS_Price'].var()
data_std_dev = stock_data['DVLPRS_Price'].std()

print("Moyenne des prix:", data_moy)
print("Médiane des prix:", data_median)
print("Variance des prix:", data_variance)
print("Écart type des prix:", data_std_dev)
```

Interpréter les résultats dans le contexte de l'analyse des données financières:

1. **Moyenne des prix**: La moyenne des prix boursiers donne une idée générale du niveau moyen des cours sur la période analysée. Si la moyenne est élevée, cela peut indiquer que les actions ont été généralement performantes.

2. **Médiane des prix**: La médiane est moins sensible aux valeurs extrêmes que la moyenne. Si la médiane est significativement différente de la moyenne, cela peut indiquer la présence de valeurs aberrantes ou d'une distribution asymétrique des prix.

3. **Variance des prix**: La variance mesure la dispersion des prix autour de la moyenne. Une variance élevée indique que les prix ont été très volatils, tandis qu'une variance faible suggère une stabilité relative des cours.

4. **Écart type des prix**: L'écart type est la racine carrée de la variance et fournit une mesure de la volatilité des prix. Un écart type élevé signifie que les prix ont fluctué considérablement, ce qui peut être un indicateur de risque pour les investisseurs.

```python
import numpy as np

# Sample data
group_a_scores = np.array([75, 88, 91, 83, 85])
group_b_scores = np.array([79, 90, 87, 88, 90])

# Calculate means
mean_a = np.mean(group_a_scores)
mean_b = np.mean(group_b_scores)
```

```python
# Calculate standard deviation
std_a = np.std(group_a_scores, ddof=1)
std_b = np.std(group_b_scores, ddof=1)
```

```python
from scipy.stats import ttest_ind

# Perform T-test
t_stat, p_val = ttest_ind(group_a_scores, group_b_scores)
print("Group A - Mean:", mean_a, "Std Dev:", std_a)
print("Group B - Mean:", mean_b, "Std Dev:", std_b)
```

```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

# Gaussian Distribution
mu, sigma = 0, 0.1  # mean and standard deviation
gaussian_dist = stats.norm(mu, sigma)
print("PDF at x=0.5:", gaussian_dist.pdf(0.5))

# Plotting Gaussian Distribution
x_gaussian = np.linspace(-0.5, 0.5, 100)
pdf_gaussian = gaussian_dist.pdf(x_gaussian)
plt.figure(figsize=(6, 4))
plt.plot(x_gaussian, pdf_gaussian)
plt.title('Gaussian Distribution\n$\mu=0$, $\sigma=0.1$')
plt.xlabel('x')
plt.ylabel('PDF')
plt.show()
```
