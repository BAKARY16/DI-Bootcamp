**Exercice 1 : Opérations matricielles**

```python
# importer les bibliothèque néccessaire
import numpy as np
```

```python
# Creation de la matirce 3x3
Matrice_3x3 = np.arange(20, 29).reshape(3, 3)
Matrice_3x3
```

```python
# Calculez le déterminant.
det = np.linalg.det(Matrice_3x3)
print("Le déterminant de la matrice 3x3 est :", det)
```

```python
# Trouvez l'inverse de la matrice
inverse = np.linalg.inv(Matrice_3x3)
print("L'inverse de la matrice 3x3 est :\n", inverse)
```

**Exercice 6 : Tests d'hypothèses de base**

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
productivity_after
# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
```

Étant donné un ensemble de données, formulez une hypothèse simple.
- Hypothèse : Le programme de formation a un effet positif sur la productivité des employés.
- hipothèse nulle (H0) : Le programme de formation n'a aucun effet sur la productivité des employés.
- hipothèse alternative (H1) : Le programme de formation a un effet positif sur la productivité des employés.

```python
# testez-la à l'aide des fonctions statistiques de base de NumPy.
productivity_diff = productivity_after - productivity_before
mean_diff = np.mean(productivity_diff)
std_diff = np.std(productivity_diff)
print(f"Mean difference in productivity: {mean_diff}")
print(f"Standard deviation of the difference: {std_diff}")
```

**Exercice 8 : Manipulation de données de séries temporelles**

```python
# Générez des données de séries chronologiques pour l'année 2023
dates = np.arange('2023-01-01', '2024-01-01', dtype='datetime64[D]')
values = np.random.rand(len(dates))
values
```

```python
# Démontrez le découpage pour les intervalles de janvier à mars
# Démontrez le découpage pour les intervalles de janvier à mars et les autres trimestres
dates_jan_mar = dates[0:90]
values_jan_mar = values[0:90]

dates_apr_jun = dates[90:181]
values_apr_jun = values[90:181]

dates_jul_sep = dates[181:273]
values_jul_sep = values[181:273]

dates_oct_dec = dates[273:365]
values_oct_dec = values[273:365]

print("Janvier à mars :", dates_jan_mar[0], "->", dates_jan_mar[-1], "(", len(values_jan_mar), "jours )")
print(values_jan_mar)

print("\nAvril à juin :", dates_apr_jun[0], "->", dates_apr_jun[-1], "(", len(values_apr_jun), "jours )")
print(values_apr_jun)

print("\nJuillet à septembre :", dates_jul_sep[0], "->", dates_jul_sep[-1], "(", len(values_jul_sep), "jours )")
print(values_jul_sep)

print("\nOctobre à décembre :", dates_oct_dec[0], "->", dates_oct_dec[-1], "(", len(values_oct_dec), "jours )")
print(values_oct_dec)
```

**Exercice 9 : Conversion de données**

```python
# Démontrer comment convertir un tableau NumPy en un DataFrame Pandas et vice versa.
import pandas as pd

# Création d'un tableau NumPy
numpy_array = np.random.rand(5, 3)
print("Tableau NumPy :\n", numpy_array)
# Conversion du tableau NumPy en DataFrame Pandas
df = pd.DataFrame(numpy_array, columns=['A', 'B', 'C'])
print("\nDataFrame Pandas :\n", df)
# Conversion du DataFrame Pandas en tableau NumPy
numpy_array_converted = df.values
print("\nTableau NumPy converti à partir du DataFrame :\n", numpy_array_converted)
```
