**Défi quotidien : Intégration de NumPy, Pandas et Matplotlib**

Vous êtes analyste de données et travaillez sur un ensemble de données météorologiques mondiales. Votre tâche consiste à analyser les tendances de température et à visualiser les résultats.

*1. Préparation des données*

```python
# Importer les bibliothèques nécessaires
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Générer des données synthétiques pour 10 villes et 12 mois
cities = [f'City {i+1}' for i in range(10)]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
# Temperatures shape: (10 villes, 12 mois)
temperatures = np.random.uniform(-5, 35, size=(10, 12))
temperatures = np.round(temperatures, 2)
# DataFrame avec villes en index et mois en colonnes
temp_df = pd.DataFrame(temperatures, index=cities, columns=months)
temp_df
```

*2. Analyse des données*

```python
# Calculer la température moyenne annuelle pour chaque ville (moyenne des 12 mois)
city_means = temp_df.mean(axis=1)
print('Température moyenne annuelle par ville :')
print(city_means.round(2))

# Identifier la ville ayant la moyenne la plus élevée et la plus basse
city_max = city_means.idxmax()
city_min = city_means.idxmin()
print()
print(f"Ville avec la température moyenne annuelle la plus élevée : {city_max} ({city_means[city_max]:.2f} °C)")
print(f"Ville avec la température moyenne annuelle la plus basse : {city_min} ({city_means[city_min]:.2f} °C)")
```

```python
# Calculer la température moyenne par mois à travers toutes les villes
month_means = temp_df.mean(axis=0)
print()
print('Température moyenne par mois (toutes villes) :')
print(month_means.round(2))
```

*3. Visualisation des données*

```python
# Tracer une courbe pour chaque ville afin de comparer les tendances mensuelles
plt.figure(figsize=(12, 7))
for city in temp_df.index:
    plt.plot(months, temp_df.loc[city], marker='o', label=city)

plt.title('Température Moyenne Mensuelle par Ville')
plt.xlabel('Mois')
plt.ylabel('Température (°C)')
plt.grid(alpha=0.3)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Résumé automatique basé sur les calculs ci-dessus
summary = (f"La ville avec la température moyenne annuelle la plus élevée est {city_max} ({city_means[city_max]:.2f} °C). ",
           f"La ville avec la température moyenne annuelle la plus basse est {city_min} ({city_means[city_min]:.2f} °C).")
print(summary)
```

Un bref rapport :
- **Méthode** : données synthétiques générées avec `np.random.uniform` pour 10 villes et 12 mois.
- **Analyse** : calculs de la moyenne annuelle par ville (`mean(axis=1)`) et moyenne mensuelle (`mean(axis=0)`).
- **Visualisation** : une courbe par ville permet de comparer les tendances saisonnières.
- **Conclusions** : voir la cellule de résumé exécutable pour les villes avec les moyennes annuelles extrêmes.
