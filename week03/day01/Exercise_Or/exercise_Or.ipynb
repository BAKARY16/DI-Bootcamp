**Exercice 1 : Minimum et maximum d'un tableau aléatoire**

```python
# Importer les bibliothèque néccessaire
import numpy as np
```

```python
# Créez un tableau 5x5 avec des valeurs aléatoires et trouvez les valeurs minimale et maximale.

tab = np.random.rand(5, 5)
tab = np.round(tab, 2)
tab
```

```python
tab_min = np.min(tab)
tab_max = np.max(tab)

print(f"Min: {tab_min}, Max: {tab_max}")
```

**Exercice 2 : Normalisation matricielle**

```python
# importer les bibliothèque néccessaire
import numpy as np
```

```python
# Normaliser une matrice aléatoire 3x3 (soustraire la moyenne et diviser par l'écart type de la matrice).
array_tab = np.random.rand(3, 3)

#Soustraire la moyenne

array_moy = array_tab.mean()
array_moy

# Diviser par l'ecart type
array_Type = np.std(array_tab)
array_Type

result = array_moy/array_Type
result
```
