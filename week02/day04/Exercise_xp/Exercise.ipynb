**Exercice 1 : Comprendre la visualisation des données**

Expliquer pourquoi la visualisation des données est importante dans l'analyse des données?

- La visualisation des données est importante dans l'analyse des données car elle permet de transformer des données brutes en représentations visuelles compréhensibles. Cela facilite l'identification de tendances, de modèles et d'anomalies qui pourraient ne pas être évidents dans un format tabulaire. De plus, les visualisations aident à communiquer efficacement les résultats de l'analyse à un public non technique, rendant les insights plus accessibles et impactants.


Décrire l'utilité d'un graphique linéaire en visualisation de données?

- L'utilité d'un graphique linéaire est de montrer l'évolution d'une variable continue au fil du temps ou par rapport à une autre variable. Il permet de visualiser les tendances, les cycles et les fluctuations, facilitant ainsi l'analyse des données temporelles ou des relations entre deux variables. Les graphiques linéaires sont particulièrement utiles pour identifier les points de changement, les pics et les creux dans les données.

**Exercice 2 : Création d’un graphique linéaire pour la variation de température**

 Créer un graphique linéaire simple à l'aide de Matplotlib qui représente les variations de température sur une semaine

```python
# import les bibliothèque néccessaire
import matplotlib.pyplot as plt
```

```python
# Utilisez une liste de valeurs de température pour chaque jour de la semaine

temp = [72, 74, 76, 80, 82, 78, 75]
```

```python
# Intitulez l'axe des x « Jour » et l'axe des y « Température (°F) ».
jours = ['Lundi', 'Mardi', 'Mercredi', 'Jeudi', 'Vendredi', 'Samedi', 'Dimanche']
plt.plot(jours, temp, marker='o')
plt.xlabel('Jour')
plt.ylabel('Température (°F)')

# Ajoutez un titre à l'intrigue
plt.title('Variation de la température au cours de la semaine')

# Afficher le graphique
plt.grid()
plt.show()
```

**Exercice 3 : Visualisation des ventes mensuelles à l’aide d’un graphique à barres**

Générer un graphique à barres à l'aide de Matplotlib pour visualiser les données de ventes mensuelles d'un magasin de détail.


```python
# import des bibliothèques nécessaires
import matplotlib.pyplot as plt
```

```python
# Créez une liste des valeurs de vente pour chaque mois
ventes = [5000, 5500, 6200, 7000, 7500]
```

```python
# Intitulez l'axe des x « Mois » et l'axe des y « Montant des ventes ($) »
plt.bar(['Jan', 'Fév', 'Mar', 'Avr', 'Mai'], ventes, color='orange')
plt.xlabel('Mois')
plt.ylabel('Montant des ventes ($)')
plt.title('Ventes mensuelles')

#Afficher le graphique
plt.grid(axis='y')
plt.show()
```

**Exercice 4 : Visualisation de la distribution des moyennes cumulatives**

 Créer un histogramme pour visualiser la distribution des moyennes cumulatives des étudiants.

 Supposons que les données CGPA soient catégorisées en plages et chargées dans un DataFrame nommé df

```python
# Importez les bibliothèques nécessaires
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('Student_Mental_health.csv')
df.head()
```

```python
# Utilisez Seaborn histplot pour créer un histogramme des catégories de CGPA
sns.histplot(df['What is your CGPA?'], bins=10, kde=True)
plt.xlabel('CGPA')
plt.ylabel('Fréquence')
plt.title('Distribution des moyennes cumulatives des étudiants')
plt.grid()
plt.show()
```

```python
# Personnalisez l'histogramme avec une couleur spécifique et ajoutez un titr
plt.title('Distribution des moyennes cumulatives des étudiants', fontsize=16, color='blue')
sns.histplot(df['What is your CGPA?'], bins=10, kde=True)
plt.xlabel('CGPA', fontsize=12, color='green')
plt.ylabel('Fréquence', fontsize=12, color='green')

# Afficher le graphiqu
plt.grid()
plt.show()
```

**Exercice 5 : Comparaison des niveaux d’anxiété selon le sexe**

Objectif : Utiliser un diagramme à barres pour comparer la proportion d'étudiants souffrant d'anxiété selon le sexe.

Aperçu du jeu de données : Le jeu de données comprend les colonnes : « Souffrez-vous d’anxiété ? » et « Choisissez votre sexe ».

```python
# Importez les bibliothèques nécessaires
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv('Student_Mental_health.csv')
data.head()
```

```python
# Utilisez Seaborn pour créer un graphique à barres comparant les niveaux d'anxiété selon le sexe à partir de l'ensemble de données data
data['Anxiety_numeric'] = data['Do you have Anxiety?'].map({'Yes': 1, 'No': 0})

sns.barplot(x='Choose your gender', y='Anxiety_numeric', data=data)
plt.xlabel('Sexe')
plt.ylabel("Proportion d'anxiété")
plt.title("Comparaison des niveaux d'anxiété selon le sexe")
plt.ylim(0, 1)
plt.grid()
plt.show()
```

```python
# Personnalisez le graphique avec une palette de couleurs appropriée et ajoutez un titre.
plt.title("Comparaison des niveaux d'anxiété selon le sexe", fontsize=16, color='purple')
sns.barplot(x='Choose your gender', y='Anxiety_numeric', data=data, palette='Set2')
plt.xlabel('Sexe', fontsize=12, color='brown') 
plt.ylabel("Proportion d'anxiété", fontsize=12, color='brown')
plt.ylim(0, 1)
plt.grid()
plt.show()
```

**Exercice 6 : Explorer la relation entre l’âge et les crises de panique**

Objectif : Créer un nuage de points pour explorer la relation entre l'âge des élèves et la survenue de crises de panique.

Aperçu du jeu de données : Le jeu de données comprend les colonnes : « Âge » et « Souffrez-vous de crises de panique ? ».

```python
# Importez les bibliothèques nécessaires
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# importer des données
datas = pd.read_csv('Student_Mental_health.csv')
datas.head()
```

```python
# Convertir les réponses aux crises de panique en valeurs numériques (par exemple, Oui=1, Non=0)
datas['Panic_numeric'] = datas['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})
datas.head()
```

```python
# Utilisez Seaborn scatterplotpour créer un nuage de points avec l'« âge » sur l'axe des x et les réponses numériques aux crises de panique sur l'axe des y.
sns.scatterplot(x='Age', y='Panic_numeric', data=datas)
plt.xlabel("Âge")
plt.ylabel("Proportion de crises de panique")

# Personnalisez le graphique pour améliorer sa lisibilité en ajoutant des étiquettes, un titre et en ajustant le style des points
plt.title("Relation entre l'âge et les crises de panique")
plt.grid()
plt.show()
```
