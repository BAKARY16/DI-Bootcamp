**Exercice 8 : Fonctions agrégées simples**

```python
# Import les bibliothèque néccessaire
import numpy as np
```

```python
# Trouvez la somme et la moyenne d'un tableau 1D donné.
list_1d = np.arange(1, 11)

# Somme de la liste
sum_list = np.sum(list_1d)
# La moynenne de la liste
moy_list = np.mean(list_1d)

print(f"Sum: {sum_list}, Average: {moy_list}")
```
