1) Qu'est-ce que l'analyse de données ?

*L'analyse de données est le processus d'examiner, de nettoyer, de transformer et de modéliser des données dans le but de découvrir des informations utiles, de tirer des conclusions et de soutenir la prise de décision. Elle implique l'utilisation de diverses techniques statistiques, algorithmiques et visuelles pour interpréter les données et en extraire des insights significatifs.*

2) Pourquoi l'analyse des données est-elle importante dans les contextes modernes ?

**Exercice 6 : Identification des types de données**

- Les rapports financiers d'une entreprise sont stockés dans un fichier Excel: ces types de données sont structurés, car ils sont organisés en lignes et en colonnes avec des étiquettes pour chaque type de données.

- Photographies téléchargées sur une plateforme de médias sociaux : ces types de données sont non structurés, car ils ne suivent pas un format pré-défini ou ne sont pas organisés de manière systématique.

- Une collection d'articles de presse sur un site web: ces données peuvent être structurées si elles sont organisées avec des métadonnées (titre, auteur, date), mais elles peuvent aussi être non structurées si elles sont simplement du texte brut sans organisation claire.

- Données d'inventaire dans une base de données relationnelle: ces données sont structurées, car elles sont organisées en tables avec des relations définies entre elles, ce qui facilite l'accès et l'analyse.

- Enregistrements d'entretiens issus d'une étude de marché: ces données sont généralement non structurées, car elles peuvent être sous forme de texte libre, d'audio ou de vidéo, et ne suivent pas un format rigide.
