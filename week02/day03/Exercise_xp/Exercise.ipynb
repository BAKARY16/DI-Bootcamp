**Exercice 2 : Statistiques descriptives**

```python
# Importer les bibliothèque néccessaire
from scipy import stats
import statistics
```

```python
# Tâche : Étant donné un ensemble de données d'exemple, calculez la moyenne, la médiane, la variance et l'écart type à l'aide de SciPy.
data = [12, 15, 13, 12, 18, 20, 22, 21]

data_mean = stats.tmean(data)
print("Moyenne:", data_mean)

data_median = statistics.median(data)
print("Médiane:", data_median)

data_variance = stats.tvar(data)
print("Variance:", data_variance)

data_std_dev = stats.tstd(data)
print("Écart type:", data_std_dev)
```

**Exercice 4 : Application du test t**

```python
# Importer les bibliothèque néccessaire
import numpy as np
from scipy import stats
```

```python
# Tâche : Effectuer un test T sur deux ensembles de données générées aléatoirement.

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_statistic, p_value = stats.ttest_ind(data1, data2)
print("Statistique t:", t_statistic)
print("P-value:", p_value)
```

**Exercice 5 : Analyse de régression linéaire**

Tâche : Étant donné un ensemble de données de prix de logements ( house_prices) et leurs tailles correspondantes ( house_sizes), utilisez la régression linéaire pour prédire le prix d'une maison en fonction de sa taille

```python
# Ensemble de donnée
house_sizes = [50, 70, 80, 100, 120] #en mètres carrés
house_prices = [150000, 200000, 210000, 250000, 280000] #en unités monétaires
```

```python
# Quelles sont la pente et l'ordonnée à l'origine de la droite de régression ?
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)
print("Pente:", slope)
```

```python
# Estimez le prix d'une maison de 90 mètres carrés
estimated_price = slope * 90 + intercept
print("Prix estimé pour une maison de 90 mètres carrés:", estimated_price)
```

Interprétez la signification de la pente dans le contexte des prix de l'immobilier.

- D'après les résultats, nous pouvons conclure que la taille de la maison a un impact significatif sur son prix, avec une pente positive indiquant que les maisons plus grandes ont tendance à être plus chères.

**Exercice 6 : Comprendre l’ANOVA**

Tâche : Trois engrais différents sont appliqués à trois groupes de plantes distincts afin de tester leur efficacité. La croissance, exprimée en centimètres, est enregistrée.

```python
from scipy import stats
import numpy as np
```

```python
# Ensemble de donnée

fertilizer_1 =  [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]
```

```python
# Effectuez un test ANOVA sur les données fournies. Quelles sont les valeurs F et p ?
f_statistic, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("Statistique F:", f_statistic)
print("P-value:", p_value)
```

D'après la valeur P, les engrais ont-ils des effets significativement différents sur la croissance des plantes ?

- D'après les résultats, nous pouvons dire que les engrais ont des effets significativement différents sur la croissance des plantes, car la valeur P est inférieure au seuil de 0,05, indiquant que les différences observées entre les groupes ne sont probablement pas dues au hasard.


Expliquez ce qui se passerait si la valeur P était supérieure à 0,05?

Dans ce cas, on pourrait dire que les engrais n'ont pas d'effets significativement différents sur la croissance des plantes, car les différences observées entre les groupes pourraient être dues au hasard, et nous ne pourrions pas rejeter l'hypothèse nulle.

**Exercice 7 : Distributions de probabilité (facultatif)**

```python
# Travailler avec une distribution binomiale : calculer les probabilités pour différents nombres de succès

n = 10  # nombre d'essais
p = 0.5  # probabilité de succès

for k in range(n + 1):
    probability = stats.binom.pmf(k, n, p)
    print(f"Probabilité d'obtenir {k} succès sur {n} essais: {probability:.4f}")

```
