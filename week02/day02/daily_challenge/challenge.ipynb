**Gestion et analyse des données en Python**

```python
# import des bibliothèques nécessaires
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
# import du jeu de donnée
data = pd.read_csv('datascience_salaries.csv')
data.head()
```

```python
# Normalisez la colonne « salaire » en utilisant la normalisation Min-Max qui met à l'échelle toutes les valeurs de salaire entre 0 et 1
scaler = MinMaxScaler()
data['Salary_Normalized'] = scaler.fit_transform(data[['salary']])
print(data[['salary', 'Salary_Normalized']].head())
```

```python
# Mettez en œuvre une réduction de dimensionnalité comme l'analyse en composantes principales (ACP) ou le t-SNE pour réduire le nombre de caractéristiques (colonnes) dans l'ensemble de données
# Ce dataset contient surtout des colonnes catégorielles, donc on encode d'abord plusieurs variables pertinentes.
features_to_encode = ['job_type', 'experience_level', 'salary_currency']
encoded = pd.get_dummies(data[features_to_encode], drop_first=True)
encoded['Salary_Normalized'] = data['Salary_Normalized']

# 1. Nettoyage : remplacer les NaN par 0 si nécessaire
imputer = SimpleImputer(strategy='constant', fill_value=0)
features_propres = imputer.fit_transform(encoded)

# 2. Normalisation des variables avant l'ACP
scaler2 = StandardScaler()
features_scalees = scaler2.fit_transform(features_propres)

# 3. Application de la PCA
pca = PCA(n_components=2)
data_reduce = pca.fit_transform(features_scalees)

# 4. Résultat
data_reduce_df = pd.DataFrame(data_reduce, columns=['PC1', 'PC2'])
print(data_reduce_df.head())
print('\nExplained variance ratio :', pca.explained_variance_ratio_)
```

```python
# Regroupez l'ensemble de données par le niveau d'expérience et calculez le salaire moyen et médian
experience_stats = data.groupby('experience_level')['salary'].agg(['mean', 'median']).rename(columns={'mean': 'salary_mean', 'median': 'salary_median'})
print('Salaires moyens et médians par niveau d\'expérience :')
print(experience_stats)
```
